q-2 Write a file using(;) as the determiner and -999 for missing values.Read it so -999 becomes NaN



```python
import pandas as pd

data = {
    "Name": ["Ishaan", "Rahul", "Aman", "Priya"],
    "Age": [20, 21, -999, 19],
    "Marks": [85, -999, 78, 92]
}

df = pd.DataFrame(data)

# Write file using ; as delimiter
df.to_csv("students.csv", sep=";", index=False, na_rep="-999")

print("File created successfully!")
```

Read a big CSV in chunks of 2000 rows and compute the mean of one coloumn per key across all chunk


```python
import pandas as pd

# Read CSV in chunks of 2000 rows
chunk_size = 2000

# Store sum and count for each key across all chunks
result = {}

for chunk in pd.read_csv("employees.csv", chunksize=chunk_size):

    # Group by key and calculate sum and count
    grouped = chunk.groupby("Department")["Salary"].agg(["sum", "count"])

    # Combine results from every chunk
    for key, row in grouped.iterrows():
        if key not in result:
            result[key] = [0, 0]

        result[key][0] += row["sum"]
        result[key][1] += row["count"]

# Calculate final mean for each key
for key, (total_sum, total_count) in result.items():
    print(key, ":", total_sum / total_count if total_count > 0 else float("nan"))
```

4.Export a Dataframe with orient="records",read it back with pd.read_json,and check equality with .equals.

```python
import pandas as pd

# Create a DataFrame
df = pd.DataFrame({
    "Name": ["Ishaan", "Rahul", "Aman"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 78]
})

# Export DataFrame to JSON
df.to_json("students.json", orient="records", indent=4)

# Read JSON back into a DataFrame
df2 = pd.read_json("students.json", orient="records")

# Check equality
print(df.equals(df2))
```

Write a short HTML table by hand and load it with read_html

```python
import pandas as pd
from io import StringIO

# Write a short HTML table by hand
html = """
<table>
    <tr>
        <th>Name</th>
        <th>Age</th>
        <th>Marks</th>
    </tr>
    <tr>
        <td>Ishaan</td>
        <td>20</td>
        <td>85</td>
    </tr>
    <tr>
        <td>Rahul</td>
        <td>21</td>
        <td>90</td>
    </tr>
    <tr>
        <td>Aman</td>
        <td>19</td>
        <td>78</td>
    </tr>
</table>
"""

# Load HTML table into a DataFrame
df = pd.read_html(StringIO(html))[0]

print(df)
```
